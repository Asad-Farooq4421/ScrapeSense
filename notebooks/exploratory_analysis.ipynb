# E-Commerce Product Scraper — Exploratory Data Analysis
This notebook analyzes the scraped and cleaned dataset stored in `output/books_data.csv`.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('../output/books_data.csv')
df.head()
```

```python
# Dataset structural overview
df.info()
df.describe()
```

```python
# Price Distribution
plt.figure(figsize=(8, 4))
plt.hist(df['price'], bins=15, color='#0f172a', edgecolor='white')
plt.title('Product Price Distribution')
plt.xlabel('Price (£)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Rating Breakdown
df['rating'].value_counts().sort_index().plot(kind='bar', color='#2563eb', figsize=(7, 4))
plt.title('Rating Distribution')
plt.xlabel('Star Rating')
plt.ylabel('Book Count')
plt.show()
```
